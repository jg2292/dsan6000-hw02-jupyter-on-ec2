# Import Libraries

```python
import boto3
from pathlib import Path
```

# Create New S3 and Move hourly data to it

```python
s3 = boto3.client("s3")

bucket = "dsan6000-jg2292"
local_dir = Path("data")

s3.create_bucket(Bucket=bucket)

for file in local_dir.glob("*.parquet"):
    s3.upload_file(
        str(file),
        bucket,
        f"wikipedia-hourly/{file.name}"
    )

print(f"Created s3://{bucket}")
print("Uploaded all Parquet files to wikipedia-hourly/")
```
